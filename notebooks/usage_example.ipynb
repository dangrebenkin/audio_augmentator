```python
from audio_augmentator import Augmentator, measure_time_for_augmentation
from datasets import load_dataset, disable_caching
from IPython.display import Audio
```

```python
# prepare test data
test_dataset = load_dataset('dangrebenkin/voxforge-ru-dataset', split='test[0:5]')
```

```python
# prepare Augmentator
augmentation_tool = Augmentator(noises_dataset='../noises_dataset')
augmentation_tool.set_augmentator_settings(snr_db=2, sample_rate=16000, 
                                           noises={"air_transport": True, "background_music": True},
                                           loop_overlay=True, random_position_overlay=False)
```

```python
# augmentations
durations, times_for_augmentations, audio = measure_time_for_augmentation(augmentation_tool, test_dataset, 16000)
durations, times_for_augmentations, audio
```

```python
Audio(audio[0]['air_transport'], rate=16000)
```

```python
Audio(audio[0]['background_music'], rate=16000)
```
